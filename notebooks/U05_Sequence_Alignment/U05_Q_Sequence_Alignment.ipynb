```
import hashlib

SALT = "U05_Q"
KEY = {'q1': 'c9513c602497', 'q2': '4d7b7ad52e5b', 'q3': 'c9513c602497', 'q4': '4d7b7ad52e5b', 'q5': '4d7b7ad52e5b', 'q6': '4d7b7ad52e5b', 'q7': 'a1608bf2eb77', 'q8': '4d7b7ad52e5b', 'q9': 'c9513c602497', 'q10': '4d7b7ad52e5b'}
REVISIT = {'q1': 'U05_C2 §3', 'q2': 'U05_C1 §4', 'q3': 'U05_C2 §4', 'q4': 'U05_C3 §4', 'q5': 'U05_C3 §4', 'q6': 'U05_C2 §5', 'q7': 'U05_C3 §3', 'q8': 'U05_C1 §3', 'q9': 'U05_C3 §3', 'q10': 'U05_C3 §3'}
results = {}

def _digest(answer):
    text = str(answer).strip().lower()
    return hashlib.sha256((SALT + text).encode()).hexdigest()[:12]

def check(qid, answer):
    """Record and report whether the answer to question qid is correct."""
    results[qid] = _digest(answer) == KEY[qid]
    print(f"{qid}: {'correct' if results[qid] else 'not yet — try again'}")

def show_score():
    done = sum(results.values())
    print(f"Score: {done} / {len(KEY)}")
    for qid in KEY:
        if not results.get(qid, False):
            print(f"  revisit {REVISIT[qid]} for {qid}")
```
